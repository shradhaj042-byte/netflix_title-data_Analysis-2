| `description` | Object | Short description or summary of the title |

### EDA (Exploratory Data Analysis)

```python
## Type-wise Content Analysis
## Bar chart
```

```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(
    kind='bar',
    figsize=(7,5),
    width=0.3,
    title='Movies vs Tv Shows'
)
plt.xlabel('Type')
plt.ylabel('Number of Titles')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Year-wise Content Analysis

This analysis shows the number of Netflix titles released in each year and helps identify the years with higher content releases.


```python
year_count = df['release_year'].value_counts().sort_index().reset_index()
year_count.columns = ['Release Year','Count']

fig = px.bar(
    year_count,
    x ='Release Year',
    y ='Count',
    title ='Year-wise Content Analysis',
    text ='Count'

)
fig.show()
```

### Rating-wise Content Analysis

This analysis shows the distribution of Netflix titles across different content ratings. It helps identify the most common ratings in the Netflix dataset.


```python
rating_count = df['rating'].value_counts().sort_index().reset_index()
rating_count.columns = ['Rating','Count']

fig = px.pie(
    rating_count,
    names= 'Rating',
    values='Count',
    title= 'Rating-wise content Analysis',


)
#fig.update_traces(
  #  textinfo='label+value+percent'
#)
fig.show()
```

## Country-wise Content Analysis

This analysis shows the countries with the highest number of Netflix titles.


Horizontal Bar Chart

```python
# Country-wise Content Analysis

country_count = df['country'].value_counts()

fig = px.bar(
    x=country_count.values,
    y=country_count.index,
    orientation='h',
    labels={
        'x': 'Number of Titles',
        'y': 'Country'
    },
    title='Countries with Netflix Content'
)

fig.show()
```
